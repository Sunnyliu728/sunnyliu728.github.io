```python
# your code here
import pandas as pd
# import plotly.express as px
```

```python
file_path = "WHR25_Data_Figure_2.1v3.xlsx"
df = pd.read_excel(file_path)
df.head()
```

```python
df = df[df["Country name"] == "Afghanistan"]
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Create figure with secondary y-axis
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add traces
fig.add_trace(
    go.Scatter(x = df["Year"], y= df["Rank"], name="Rank over time"),
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x = df["Year"], y = df["Life evaluation (3-year average)"], name="Happiness over time"),
    secondary_y=True,
)

# Add figure title
fig.update_layout(
    title_text="Afghanistan's Happiness Score and Rank over Time"
)

# Set x-axis title
fig.update_xaxes(title_text="Year")

# Set y-axes titles
fig.update_yaxes(title_text="Rank over time", secondary_y=False)
fig.update_yaxes(title_text="Happiness over time", secondary_y=True)

fig.show()
```
