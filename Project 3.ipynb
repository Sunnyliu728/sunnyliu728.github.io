## Life Evaluation and Log GDP per Capita: What Drives Happiness Across Countries?
Understanding what shapes national well-being is one of the central questions in economics, psychology, and global policy research.
In this post, I explore the relationship between GDP per capita and Life Evaluation scores across countries using 2023 data.

While GDP is not a perfect proxy for quality of life, it remains one of the most widely used indicators of economic development. I assume that Life Score is positively related to the GDP Per Capita.

- **Dataset(s) to be used:** World Hapiness Report(https://www.worldhappiness.report/) and GPD Per Capita (https://data.worldbank.org/indicator/NY.GDP.PCAP.CD)
- **Analysis question:** Does GDP per Capita drives happiness across countries?
- **Columns that will (likely) be used:**
  - [Dataset 1][Year]
  - [Dataset 1][Country name]
  - [Dataset 1][Life evaluation]
  - [Dataset 2][Country Name]
  - [Dataset 2][2023]
- (If you're using multiple datasets) **Columns to be used to merge/join them:**
  - [Dataset 1] [Country name]
  - [Dataset 2] [Country Name]
- **Hypothesis**: GDP per Capita is positively correlated with happiness

```python
import pandas as pd
import numpy as np
import plotly.express as px
from IPython.display import HTML
```

## Exploring the Data Through Two Maps
## First Database: World Happiness Report: Life Evaluation (2023) from World Happniess Report(https://www.worldhappiness.report/)

Map 1 — Life Evaluation Scores (2023)
This map shows how people in different countries rate their overall life satisfaction. High-scoring regions include Finland, Australia, and Canada, while lower scores are concentrated in regions facing political instability or economic hardship.

```python
happiness = pd.read_csv("Happiness by Country.csv")
happiness.head()
```

As to create a map, the relevant columns would be Year, Country name and Lfe evaluation.

```python
# Filter the "Year" column and only keep Year == 2023
hap_2023 = happiness[happiness["Year"] == 2023][[
    "Country name",
    "Life evaluation (3-year average)"
]]

hap_2023.head()
```

To prepare the data, I renamed the relevant columns so they match Plotly’s expected inputs:
 - Country name → Country

 - Life evaluation → LifeScore

This makes it easier to merge with other datasets later and ensures the map can correctly match country names.

```python
# Rename the columns
hap_2023 = hap_2023.rename(columns={
    "Country name": "Country",
    "Life evaluation (3-year average)": "LifeScore"
})

hap_2023.head()
```

Next, I used px.choropleth() to draw a world map where each country is shaded according to its average life evaluation score in 2023. Darker colors represent higher life satisfaction.

```python
# Draw the map
fig_1 = px.choropleth(
    hap_2023,
    locations = "Country",
    locationmode = "country names",      
    color = "LifeScore",
    color_continuous_scale = "Blues",
    title = "World Happiness Score (Life Evaluation), 2023"
)

HTML(fig_1.to_html(include_plotlyjs="cdn", full_html=False))
```

## Second Database: Gdp Per Capita from World Bank(https://data.worldbank.org/indicator/NY.GDP.PCAP.CD)

Map 2 — Log GDP per Capita (2023)
This map highlights global economic differences. Wealthier regions such as North America, Western Europe, and parts of East Asia appear in darker shades, while lower-income regions remain lighter.

```python
Gdp = pd.read_csv("GDP Per Capita by Country.csv")
Gdp.head()
```

As to create a map, the relevant columns would be Country Name and 2023.

```python
Gdp_2023 = Gdp[[
    "Country Name",
    "2023"
]]

Gdp_2023.head()
```

```python
# Rename the columns
Gdp_2023 = Gdp_2023.rename(columns={
    "Country Name": "Country",
    "2023": "Gdp 2023"
})
Gdp_2023["log_gdp"] = np.log10(Gdp_2023["Gdp 2023"])

Gdp_2023.head()
```

```python
# Draw the map
fig_2 = px.choropleth(
    Gdp_2023,
    locations = "Country",              # Country Name
    locationmode = "country names",     # Match via country names
    color = "log_gdp",                  # Use color to denote the log_gdp
    color_continuous_scale = "Blues",   
    title = "GDP Per Capita, 2023"
)

HTML(fig_2.to_html(include_plotlyjs="cdn", full_html=False))
```

# Interpretation of the two maps
By comparing the two maps side by side, an intuitive pattern emerges:

Countries with higher GDP per capita tend to also report higher life evaluation scores, although there are notable exceptions.

This initial visual exploration motivated the next step—creating a scatter plot to examine the relationship more directly and quantify the trend across nations.

# Creating a Scatter Plot
First, I need to merge two databases together, using the same country column.

```python
merged = pd.merge(hap_2023, Gdp_2023, on="Country", how="left")

merged.head()
```

Use Plotly to create an interactive scatter plot with a regression line:

 - Blue points represent all countries

 - The red line is the best-fit trend

 - Only the selected countries receive text labels

This keeps the figure readable while emphasizing meaningful comparisons.

Highlighting Selected Countries

 - To avoid clutter, we manually select several countries that illustrate interesting contrasts:

```python
selected_countries = ["Finland", "Denmark", "United States", "China", "India", "Singapore"]

subset = merged[merged["Country"].isin(selected_countries)]

fig_3 = px.scatter(
    merged,
    x = "log_gdp",
    y = "LifeScore",
    trendline = "ols",
    trendline_color_override = "red",
    title = "Life Evaluation vs Log GDP per Capita (2023)"
)

# Label the countries I selected
for i, row in subset.iterrows():
    fig_3.add_annotation(
        x = row["log_gdp"],
        y = row["LifeScore"] + 0.12,
        text = row["Country"],
        showarrow = False,
        font = dict(size = 11, color = "black")
    )

HTML(fig_3.to_html(include_plotlyjs="cdn", full_html=False))
```

 # Interpreting the Relationship Between Log GDP per Capita and Life Evaluation

The scatter plot shows a clear positive relationship between a country’s economic prosperity (measured as log GDP per capita) and its average life evaluation score. As log GDP increases, countries tend to report higher levels of subjective well-being. The assumption is correct.

A few selected countries illustrate this pattern:

 - Finland and Denmark, two of the wealthiest and happiest countries, appear in the top-right corner with both high GDP and very high life evaluations.

 - The United States and Singapore also fall along the upper end of the trend, though their life evaluation scores are slightly lower than the Scandinavian countries despite comparable GDP levels.

 - China sits around the middle of the plot, reflecting its rapidly rising GDP and moderate life satisfaction.

Overall, the red trend line confirms the pattern:

Higher income levels are strongly associated with higher life satisfaction, although there is still considerable variation among countries with similar GDP levels.

This suggests that while economic development plays a major role in shaping well-being, other social, cultural, and institutional factors also matter—especially among higher-income nations where the differences in life satisfaction become more nuanced.
